# 09 — Build a catalog from the stored base catalog

The short version of notebook 07. The base flux catalog already exists (step 1 of notebook 07, or
`romanmock base`): galaxies with a line above 1.0e-16 erg/s/cm² in the grism bandpass, with the flux of every
line. Building a catalog from it takes four things: where the base catalog is, where the output goes, the flux
limits, and the columns to keep. It runs on a login node in about two minutes (4 workers, 2.4 GB each).

````python
import os
os.environ.setdefault("JAX_PLATFORMS", "cpu")   # diffsky imports JAX; avoid CUDA probing on login nodes
import warnings
warnings.filterwarnings("ignore", message="Version checking")

import numpy as np
import matplotlib.pyplot as plt
import opencosmo as oc

from romanmock import plotting
plotting.use_style()

CATALOG = "/global/cfs/cdirs/m4943/Simulations/sdss_feniks_hizels_260710_09_07_2026"
WORK = os.path.join(os.environ.get("PSCRATCH", "/tmp"), "romanmock_demo")   # outputs of the tutorials
os.makedirs(WORK, exist_ok=True)
````

## 1. Settings

````python
from romanmock import RomanMockConfig, RomanMockPipeline
from romanmock.io import read_manifest, suggest_n_workers

BASE_DIR = os.path.join(WORK, "two_step", "base")       # written by notebook 07 / `romanmock base`

FLUX_LIMIT = 1.5e-16          # primary limit [erg/s/cm²]: at least one line of the galaxy above it
SECONDARY_LIMIT = 4.0e-17     # limit for the other lines of such a galaxy (None: same as the primary limit)
COLUMNS = ["gal_id", "ra", "dec", "redshift_true"]      # copied from the base catalog
SEED = 2026

OUT_DIR = os.path.join(WORK, "two_step", f"catalog_F{FLUX_LIMIT:.1e}_S{SECONDARY_LIMIT:.1e}")

base = read_manifest(BASE_DIR, "romanmock_base")
assert base is not None, f"no base catalog in {BASE_DIR}: run notebook 07 (step 1) or `romanmock base` first"
sens = base["base"]["sensitivity"]
print(f"base catalog: {sum(v['n_galaxies'] for v in base['slices'].values()):,} galaxies in {len(base['slices'])} slices")
print(f"  limits of the base catalog: primary {sens['flux_limit']:.1e}, bandpass {sens['wave_range']} Å")
print(f"  columns available: {base['base']['columns']} + redshift_rsd + line fluxes")
````

## 2. Build the catalog

`source: "base"` tells the pipeline to read the base catalog instead of the input mock. The primary limit must
not be deeper than the one of the base catalog; the secondary limit is free. The confusion matrix is the
package default unless a `confusion` block is given.

````python
import time

cfg = RomanMockConfig.from_dict({
    "source": "base",
    "base": {"dir": BASE_DIR},
    "output": {"dir": OUT_DIR, "passthrough_columns": COLUMNS},
    "sensitivity": {"type": "constant", "flux_limit": FLUX_LIMIT, "secondary_flux_limit": SECONDARY_LIMIT,
                    "wave_range": [10000.0, 19300.0]},
    "seed": SEED,
})
pipe = RomanMockPipeline(cfg)
n_workers = suggest_n_workers(max_workers=8, per_worker_gb=2.5)
t0 = time.time()
files = pipe.run(n_workers=n_workers, verbose=False)      # finished slices are reused on a second call
cfg.to_yaml(os.path.join(OUT_DIR, "config.yaml"))
print(f"{len(files)} files in {OUT_DIR}\n{n_workers} workers, {time.time() - t0:.0f} s")
````

## 3. Use the catalog

The slices open together as one `opencosmo.Lightcone`. `redshift_assigned` is the measured redshift,
`id_outcome` tells how it was obtained (1 multi-line, 2 single line identified correctly, 3 single line
misidentified = interloper, 4 failed), and `line_bitmask` lists the detected lines.

````python
from astropy.table import Table
from romanmock.diagnostics import outcome_summary

catalog = pipe.open_parent()
print(f"{len(catalog):,} galaxies, z_range {catalog.z_range}")
print("columns:", catalog.columns)
print(pipe.line_set.describe())

data = catalog.select("redshift_assigned", "id_outcome", "line_assigned", "n_det").get_data("numpy")
for name, n in outcome_summary(data).items():
    print(f"  {name:<15s} {n:>12,}  ({100 * n / len(catalog):6.2f}%)")

show = ["gal_id", "ra", "dec", "redshift_true", "redshift_rsd", "redshift_assigned", "id_outcome", "line_bitmask", "n_det"]
first = oc.open(files[len(files) // 2]).select(*show).get_data("numpy")
t = Table({k: first[k][:8] for k in show})
for c in ("ra", "dec", "redshift_true", "redshift_rsd", "redshift_assigned"):
    t[c].format = ".5f"
t
````

````python
from romanmock.diagnostics import dndz
from romanmock.selection import sky_area_deg2

area = sky_area_deg2(catalog)
z_edges = np.linspace(0.0, 3.0, 61); zc = 0.5 * (z_edges[1:] + z_edges[:-1])
fig, ax = plt.subplots(figsize=(8.5, 4.2))
for name in ["Ha", "OIII_5007", "SIII_9532"]:
    sel = (data["line_assigned"] == pipe.line_set.bit(name)) & np.isfinite(data["redshift_assigned"])
    ax.step(zc, dndz(data["redshift_assigned"][sel], z_edges, area), where="mid", color=plotting.feature_color(name),
            label=f"{name} ({sel.sum():,})")
ax.set_yscale("log"); ax.set_ylim(30, None); ax.set_xlabel("redshift_assigned"); ax.set_ylabel("dN/dz per deg²")
ax.set_title(f"Galaxies by assigned line, F > {FLUX_LIMIT:.1e} (other lines > {SECONDARY_LIMIT:.1e}), {area:.0f} deg²")
ax.legend(fontsize=8); plt.show()
````

## The same from the command line

```bash
romanmock assign cfg.yaml --flux-limit 1.5e-16 --secondary-flux-limit 4e-17 \
                 --output-dir $PSCRATCH/my_catalog --workers 4
```

where `cfg.yaml` contains `base: {dir: <base catalog>}` and, if needed, `output: {passthrough_columns: [...]}`,
`confusion: {...}` and `seed`. The effective configuration is written to `<output-dir>/config.yaml`; use it with
`romanmock summary` and `romanmock select` (analysis samples, see notebook 07). To change the limits, change
`FLUX_LIMIT` / `SECONDARY_LIMIT` and run again: each pair of limits gets its own output directory.
