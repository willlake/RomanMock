# 05 — Line fluxes as a function of observed wavelength

Notebook 04 shows the flux of each line as a function of **redshift**. The grism sees **observed wavelength**
λ_obs = λ_rest (1 + z), so this notebook redraws the same distributions against λ_obs inside the grism bandpass
(10 000–19 300 Å): the quantiles of the flux of each line, and the mix of lines that exceed a flux threshold in
twelve slices of the bandpass. The lines are the seven that matter for the catalog; [O III] λ4363 and [S III] λ6312
are too faint to be detected and are left out.

The input is `census/flux_distribution.npz` (`romanmock flux-distribution`, part of `scripts/production.sh`):
log₁₀ F of every galaxy of the lightcone, with no flux cut, in bins of 0.02 dex per redshift bin of 0.01.
The wavelength of a redshift bin is λ_rest (1 + z) at the bin centre; a bin of 0.01 in z is 37–95 Å wide
in λ_obs depending on the line.

```python
import os
os.environ.setdefault("JAX_PLATFORMS", "cpu")      # diffsky imports JAX; keep it off the GPU

import math
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
from matplotlib.patches import Patch
from romanmock import plotting
from romanmock.fluxdist import FluxDistribution
from romanmock.lines import get_line_set
from romanmock.roman import GRISM_WAVE_RANGE_ANGSTROM as WAVE_RANGE

plotting.use_style()

MOCK = "sdss_feniks_hizels_260710_09_07_2026"                          # version of the diffsky mock
ROOT = f"/global/cfs/cdirs/m4943/Simulations/RomanMocks/{MOCK}"        # products of romanmock for this mock

dist = FluxDistribution.load(f"{ROOT}/census/flux_distribution.npz")
line_set = get_line_set("roman_transitions")
z = dist.z

LINES = ["Ha", "OIII_5007", "OIII_4959", "Hb", "OII_3727", "SIII_9532", "SIII_9069"]     # [O III] 4363 and [S III] 6312 left out
LABEL = {"Ha": "Hα", "Hb": "Hβ", "OIII_5007": "[O III] λ5007", "OIII_4959": "[O III] λ4959", "OII_3727": "[O II] λ3727",
         "SIII_9532": "[S III] λ9532", "SIII_9069": "[S III] λ9069"}
rest = {line: line_set.rest_waves[line_set.bit(line)] for line in LINES}
wave = {line: rest[line] * (1 + z) for line in LINES}                   # observed wavelength of every redshift bin
sci = lambda x: rf"${float(f'{x:.1e}'.split('e')[0]):g}\times10^{{{int(f'{x:.1e}'.split('e')[1])}}}$"

def in_band(line, lo=WAVE_RANGE[0], hi=WAVE_RANGE[1]):
    '''Redshift bins whose observed wavelength of `line` falls in [lo, hi).'''
    return (wave[line] >= lo) & (wave[line] < hi)

print(f"{dist.n_gal.sum():,} galaxies, {dist.z_edges[0]:.1f} < z < {dist.z_edges[-1]:.1f}, bins of {dist.z_edges[1] - dist.z_edges[0]:.2f} in z")
for line in LINES:
    sel = in_band(line)
    print(f"{LABEL[line]:<14s} {rest[line]:7.1f} Å   in the bandpass for {z[sel].min():.2f} < z < {z[sel].max():.2f}  ({sel.sum()} redshift bins)")
```

## Flux quantiles against observed wavelength

For each line, the median flux (solid line) and the central 1σ and 2σ intervals of a Gaussian: the 15.9–84.1 %
quantiles (dark band) and the 2.3–97.7 % quantiles (light band) of **all galaxies** whose line falls at that
observed wavelength, including the galaxies without the line (zero flux), which sit at the faint end. Where a
lower quantile reaches the floor of the histograms (10⁻²¹) it is clipped there and leaves the plotted range.
The horizontal lines are the thresholds in use, 4×10⁻¹⁷ (dotted) and 1.5×10⁻¹⁶ (solid). The last panel repeats
the medians of all lines.

The sawtooth is the one of notebook 04: the lightcone is built from snapshots, and the flux falls with the
luminosity distance across a slice while the luminosities stay those of one snapshot.

```python
gauss = lambda k: 0.5 * (1 + math.erf(k / math.sqrt(2)))               # Gaussian cumulative probability at k sigma
Q = {1: (gauss(-1), gauss(1)), 2: (gauss(-2), gauss(2))}
THRESHOLDS = {4e-17: ":", 1.5e-16: "-"}                                 # erg/s/cm^2: line style of the marker

fig, axes = plt.subplots(2, 4, figsize=(17, 7.6), sharex=True, sharey=True, constrained_layout=True)
for ax, line in zip(axes.ravel(), LINES):
    sel = in_band(line)
    x, col = wave[line][sel], plotting.feature_color(line)
    q = {p: dist.quantile(line, p)[sel] for p in (0.5, *Q[1], *Q[2])}
    ax.fill_between(x, q[Q[2][0]], q[Q[2][1]], color=col, alpha=0.18, lw=0)
    ax.fill_between(x, q[Q[1][0]], q[Q[1][1]], color=col, alpha=0.45, lw=0)
    ax.plot(x, q[0.5], color=col, lw=1.8)
    for value, ls in THRESHOLDS.items():
        ax.axhline(value, color=plotting.TEXT_SECONDARY, lw=0.9, ls=ls)
    ax.set_title(f"{LABEL[line]}  ({rest[line]:.0f} Å, {z[sel].min():.2f} < z < {z[sel].max():.2f})", loc="left")
ax = axes.ravel()[-1]
for line in LINES:
    sel = in_band(line)
    ax.plot(wave[line][sel], dist.quantile(line, 0.5)[sel], color=plotting.feature_color(line), lw=1.8, label=LABEL[line])
for value, ls in THRESHOLDS.items():
    ax.axhline(value, color=plotting.TEXT_SECONDARY, lw=0.9, ls=ls)
ax.set_title("medians of all lines", loc="left")
ax.legend(frameon=False, fontsize=8, loc="upper right", ncol=2)
for ax in axes.ravel():
    ax.set_yscale("log")
    ax.set_ylim(1e-19, 1e-12)
    ax.set_xlim(*WAVE_RANGE)
for ax in axes[:, 0]:
    ax.set_ylabel("line flux [erg s$^{-1}$ cm$^{-2}$]")
for ax in axes[-1]:
    ax.set_xlabel("observed wavelength [Å]")
handles = [Line2D([], [], color=plotting.TEXT_PRIMARY, lw=1.8, label="median"),
           Patch(color=plotting.TEXT_PRIMARY, alpha=0.45, label="1σ: 15.9 % – 84.1 %"),
           Patch(color=plotting.TEXT_PRIMARY, alpha=0.18, label="2σ: 2.3 % – 97.7 %")]
handles += [Line2D([], [], color=plotting.TEXT_SECONDARY, lw=0.9, ls=ls, label=f"threshold {sci(v)}") for v, ls in THRESHOLDS.items()]
fig.legend(handles=handles, loc="outside lower center", ncol=5, frameon=False)
plt.show()
```

## Which lines exceed a threshold, per slice of the bandpass

The bandpass is cut into twelve ranges of 775 Å. In each range and for each flux threshold F, the lines above
F are counted over all redshift bins whose observed wavelength falls in the range, and each line is shown as its
**share of all the lines above F in that range**: the shares of the seven lines sum to one. The vertical line is
the primary threshold 1.5×10⁻¹⁶. Shares are drawn only where at least 100 lines exceed the threshold, which is
why the curves end before 10⁻¹².

At the faint end every line present in the range contributes, with the fainter lines (Hβ, [O III] λ4959, [O II])
at 10–20 % each. Around the threshold in use Hα carries half to two thirds of the lines in every range, with
[O III] λ5007 second at the blue end of the bandpass and [S III] λ9532 second at the red end. Above a few 10⁻¹⁶
the bright lines of the nearby galaxies take over: [S III] λ9532 and λ9069 at z < 1, whose flux is large simply
because the luminosity distance is small; at the red end of the bandpass Hα returns above 10⁻¹⁵, where
[S III] λ9532 comes from z ≈ 1 and no longer wins. The table below gives the shares at the threshold in use.

```python
N_PANELS = 12
F_LINE = 1.5e-16                                                         # erg/s/cm^2, the vertical marker
logf = np.linspace(-17, -12, 201)                                        # thresholds 1e-17 .. 1e-12
edges = np.linspace(WAVE_RANGE[0], WAVE_RANGE[1], N_PANELS + 1)

def number_above(line, lo, hi):
    '''Galaxies with `line` observed in [lo, hi) and brighter than each threshold of `logf`.'''
    i, sel = dist.index(line), in_band(line, lo, hi)
    counts = dist.counts[i][sel].sum(axis=0)                             # log-flux histogram summed over the redshift bins
    cum = np.concatenate([np.cumsum(counts[::-1])[::-1], [0]]) + dist.above[i][sel].sum()   # at every flux edge
    return np.interp(logf, dist.logf_edges, cum, left=cum[0], right=cum[-1])   # linear inside a bin, as FluxDistribution.fraction_above

fig, axes = plt.subplots(3, 4, figsize=(17, 10.5), sharex=True, sharey=True, constrained_layout=True)
shares_at_F = {}
for ax, lo, hi in zip(axes.ravel(), edges[:-1], edges[1:]):
    n = {line: number_above(line, lo, hi) for line in LINES}
    total = sum(n.values())
    with np.errstate(invalid="ignore", divide="ignore"):
        share = {line: np.where(total >= 100, n[line] / total, np.nan) for line in LINES}
    for line in LINES:
        if np.nanmax(share[line]) > 0:
            ax.plot(10.0 ** logf, share[line], color=plotting.feature_color(line), lw=1.8, label=LABEL[line])
    ax.axvline(F_LINE, color=plotting.TEXT_SECONDARY, lw=0.9, ls="-")
    ax.set_xscale("log")
    ax.set_xlim(1e-17, 1e-12)
    ax.set_ylim(0, 1)
    ax.set_title(f"{lo:.0f} – {hi:.0f} Å", loc="left")
    shares_at_F[(lo, hi)] = {line: np.interp(np.log10(F_LINE), logf, share[line]) for line in LINES}
for ax in axes[:, 0]:
    ax.set_ylabel("share of the lines above the threshold")
for ax in axes[-1]:
    ax.set_xlabel("flux threshold [erg s$^{-1}$ cm$^{-2}$]")
handles = [Line2D([], [], color=plotting.feature_color(line), lw=1.8, label=LABEL[line]) for line in LINES]
handles.append(Line2D([], [], color=plotting.TEXT_SECONDARY, lw=0.9, label=f"threshold {sci(F_LINE)}"))
fig.legend(handles=handles, loc="outside lower center", ncol=8, frameon=False)
plt.show()

print(f"share of each line among the lines above {F_LINE:.1e} erg/s/cm^2, per wavelength range")
print(f"{'range [Å]':<16s}" + "".join(f"{LABEL[l]:>15s}" for l in LINES))
for (lo, hi), s in shares_at_F.items():
    print(f"{lo:6.0f} – {hi:5.0f}  " + "".join(f"{s[l]:15.3f}" if np.isfinite(s[l]) else f"{'—':>15s}" for l in LINES))
```
