# 01 — dN/dz and the flux limits

Number density of the emission-line galaxies of the mock as a function of redshift, for the flux limits of the
survey. A galaxy is detected when one of its lines exceeds the **primary** limit inside the grism bandpass
(1.00–1.93 μm). Its other lines then count as detected above a lower **secondary** limit.

For a line X, with primary limit P and secondary limit S:

| class | definition |
|---|---|
| all | X is detected: X > P, or X > S while another line of the galaxy is above P |
| single | X > P and no other line above S: the redshift rests on the identification of one line |
| multi | X is detected together with at least one other line |

The counts contain the flux, bandpass and redshift cuts only: no line misidentification and no downsampling.
The densities are per unit of the sky area `AREA_DEG2`, which is estimated when it is not given.
The counts are read from the census written by `sbatch scripts/production.sh` (`romanmock census`).

```python
import os
os.environ.setdefault("JAX_PLATFORMS", "cpu")      # diffsky imports JAX; keep it off the GPU

import numpy as np
import matplotlib.pyplot as plt
from romanmock import plotting

plotting.use_style()

MOCK = "sdss_feniks_hizels_260710_09_07_2026"                          # version of the diffsky mock
ROOT = f"/global/cfs/cdirs/m4943/Simulations/RomanMocks/{MOCK}"        # products of romanmock for this mock
```

```python
from romanmock import BaseCatalog
from romanmock.census import LineCensus

census = LineCensus.load(f"{ROOT}/census/dndz_census_z0p5_3p0.npz")

AREA_DEG2 = None                            # sky area [deg^2] if you know it; None: estimated
base = BaseCatalog(f"{ROOT}/base", area_deg2=AREA_DEG2)
area = base.area_deg2

PRIMARY = [1.0e-16, 1.5e-16]                # erg/s/cm^2
SECONDARY = [None, 8.0e-17, 4.0e-17]        # None: one limit for every line
LINES = ["Ha", "OIII_5007", "Hb", "OII_3727"]

z_edges = census.z_edges
z = 0.5 * (z_edges[1:] + z_edges[:-1])
color = dict(zip(SECONDARY, plotting.CATEGORICAL))
flux = lambda x: f"{x:.1e}".replace(".0e", "e")                                # 1e-16, 1.5e-16, 8e-17
label = lambda primary, s: f"secondary limit {flux(primary if s is None else s)}"   # None: the primary limit

print(f"{len(census.steps)} slices, {z_edges[0]:.1f} < z < {z_edges[-1]:.1f}, {area:.2f} deg2 ({base.area_source})")
print("limits (primary, secondary):", census.limits)
```

## dN/dz

Rows: primary limit. Columns: line. Colours: secondary limit; with no secondary limit it equals the primary limit. Solid: all galaxies with the line detected;
dashed: single-line; dotted: multi-line.

```python
style = {"all": "-", "single": "--", "multi": ":"}
fig, axes = plt.subplots(len(PRIMARY), len(LINES), figsize=(17, 7.6), sharex=True, sharey=True, constrained_layout=True)
for row, primary in zip(axes, PRIMARY):
    for ax, line in zip(row, LINES):
        for s in SECONDARY:
            for which, ls in style.items():
                ax.step(z, census.dndz(line, primary, area, which, secondary=s), where="mid", color=color[s], ls=ls,
                        lw=1.7 if which == "all" else 1.2, label=f"{label(primary, s)}, {which}")
        ax.set_yscale("log")
        ax.set_ylim(10, 5e4)
        ax.set_xlim(z_edges[0], z_edges[-1])
        ax.set_title(f"{line}, primary limit {primary:.1e}", loc="left")
    row[-1].legend(*row[0].get_legend_handles_labels(), loc="center left", bbox_to_anchor=(1.02, 0.5), frameon=False, fontsize=8)
    row[0].set_ylabel("dN/dz per deg$^2$")
for ax in axes[-1]:
    ax.set_xlabel("redshift")
plt.show()
```

## Comoving number density

The same counts per comoving volume, for Hα and [O III] λ5007, in the cosmology of the mock (h = 0.6766,
Ω_m = 0.310). The volume of a redshift bin is the comoving volume between its edges over the sky area.

Two thresholds define the sample, in erg s⁻¹ cm⁻²:

- **Galaxy-selection threshold** $F_{\rm sel}$ (rows): a galaxy enters the sample if at least one of its lines
  inside the grism bandpass exceeds $F_{\rm sel}$.
- **Line-detection threshold** $F_{\rm line}$ (colours): within this selected sample, the line X named in a
  panel counts as detected if it lies inside the bandpass and exceeds $F_{\rm line}$. X itself need not exceed
  $F_{\rm sel}$: another line of the galaxy can be the one that selects it.

Three detection classes (line styles):

- **All galaxies with X detected** (solid);
- **Only X detected** (dashed): no other in-band line exceeds $F_{\rm line}$, so X is the selecting line and
  does exceed $F_{\rm sel}$;
- **X + other line(s) detected** (dotted): at least one additional detection.

The solid curve is the sum of the other two.

```python
import glob
import opencosmo as oc
from astropy import units as u
from matplotlib.lines import Line2D

cosmo = oc.open(sorted(glob.glob(f"/global/cfs/cdirs/m4943/Simulations/{MOCK}/lc_cores-*.diffsky_gals.hdf5"))[0]).cosmology
sky_fraction = area / (4 * np.pi * np.degrees(1.0) ** 2)
volume = np.diff(cosmo.comoving_volume(z_edges).to(u.Mpc**3).value) * cosmo.h**3 * sky_fraction   # (Mpc/h)^3 per bin

def number_density(line, primary, which, secondary):
    """Galaxies with the line detected per comoving volume [h^3 Mpc^-3], per redshift bin."""
    return census.hist(line, primary, which, secondary=secondary) / volume

def sci(x):
    """1.5e-16 -> 1.5\times10^{-16} (mathtext)."""
    m, e = f"{x:.1e}".split("e")
    return rf"{m}\times10^{{{int(e)}}}"

DENSITY_LINES = {"Ha": "Hα", "OIII_5007": "[O III] λ5007"}
fig, axes = plt.subplots(len(PRIMARY), len(DENSITY_LINES), figsize=(12.5, 7.6), sharex=True, sharey=True, constrained_layout=True)
for row, primary in zip(axes, PRIMARY):
    for ax, (line, name) in zip(row, DENSITY_LINES.items()):
        for s in SECONDARY:
            for which, ls in style.items():
                ax.step(z, number_density(line, primary, which, s), where="mid", color=color[s], ls=ls,
                        lw=1.7 if which == "all" else 1.2)
        ax.set_yscale("log")
        ax.set_ylim(1e-6, 1e-2)
        ax.set_xlim(z_edges[0], z_edges[-1])
        ax.set_title(rf"{name} detections — $F_{{\rm primary}} = {sci(primary)}$", loc="left")
    row[0].set_ylabel("n [$h^3$ Mpc$^{-3}$]")
for ax in axes[-1]:
    ax.set_xlabel("redshift")

thresholds = [Line2D([], [], color=color[s], lw=1.7,
                     label=r"$F_{\rm secondary} = F_{\rm primary}$" if s is None else rf"$F_{{\rm secondary}} = {sci(s)}$") for s in SECONDARY]
classes = [Line2D([], [], color=plotting.TEXT_PRIMARY, ls=ls, lw=1.4, label=text) for ls, text in
           [("-", "All galaxies with X detected"), ("--", "Only X detected"), (":", "X + other line(s) detected")]]
fig.legend(handles=thresholds, title="Line-detection threshold\n[erg s$^{-1}$ cm$^{-2}$]", loc="outside right upper",
           frameon=False, fontsize=9, title_fontsize=9, alignment="left")
fig.legend(handles=classes, title="Detection class\n(X: the line of the panel)", loc="outside right lower",
           frameon=False, fontsize=9, title_fontsize=9, alignment="left")
plt.show()

print(f"{cosmo.name or 'mock cosmology'}: H0 = {cosmo.H0.value:.2f}, Om0 = {cosmo.Om0:.4f}")
print(f"comoving volume of the census, 0.5 < z < 3.0: {volume.sum() / 1e9:.2f} (Gpc/h)^3")
```

## Single-line fraction of each line's emitters

Fraction of the galaxies with the line detected that have no other detected line.

```python
fig, axes = plt.subplots(1, len(PRIMARY), figsize=(13, 4.4), sharey=True, constrained_layout=True)
for ax, primary in zip(axes, PRIMARY):
    for line in ["Ha", "OIII_5007"]:
        for s, ls in zip(SECONDARY, ["-", "--", ":"]):
            ax.step(z, census.single_fraction(line, primary, secondary=s), where="mid", color=plotting.feature_color(line),
                    ls=ls, label=f"{line}, {label(primary, s)}")
    ax.set_ylim(0, 1)
    ax.set_xlim(z_edges[0], z_edges[-1])
    ax.set_xlabel("redshift")
    ax.set_title(f"primary limit {primary:.1e}", loc="left")
    ax.legend(frameon=False, fontsize=8, loc="upper left")
axes[0].set_ylabel("single-line fraction")
plt.show()
```

## Findings

- **Hα** is almost always the brightest line where it is inside the bandpass (z < 1.94), so its total counts do
  not depend on the secondary limit. Its single-line fraction does: with 4e-17 nearly every Hα emitter has a
  second line.
- **[O III] 5007** at z > 1.94, where Hα has left the bandpass, is single-line without a secondary limit and
  multi-line with 4e-17, because [O III] 4959 (flux ratio 1 : 3.04) and Hβ are then detected.
- A secondary limit of 4e-17 is 2.6σ for a 6.5σ limit of 1e-16. The counts assume that such a line is usable
  once another line has anchored the galaxy; they contain no noise.
