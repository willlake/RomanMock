# 01 — dN/dz and the flux limits

Number density of the emission-line galaxies of the mock as a function of redshift, for the flux limits of the
survey. A galaxy is detected when one of its lines exceeds the **primary** limit inside the grism bandpass
(1.00–1.93 μm). Its other lines then count as detected above a lower **secondary** limit.

For a line X, with primary limit P and secondary limit S:

| class | definition |
|---|---|
| all | X is detected: X > P, or X > S while another line of the galaxy is above P |
| single | X > P and no other line above S: the redshift rests on the identification of one line |
| multi | X is detected together with at least one other line |

The counts contain the flux, bandpass and redshift cuts only: no line misidentification and no downsampling.
They are read from the census written by `sbatch scripts/production.sh` (`romanmock census`).

```python
import os
os.environ.setdefault("JAX_PLATFORMS", "cpu")      # diffsky imports JAX; keep it off the GPU

import numpy as np
import matplotlib.pyplot as plt
from romanmock import plotting

plotting.use_style()

MOCK = "sdss_feniks_hizels_260710_09_07_2026"                          # version of the diffsky mock
ROOT = f"/global/cfs/cdirs/m4943/Simulations/RomanMocks/{MOCK}"        # products of romanmock for this mock
```

```python
from romanmock import BaseCatalog
from romanmock.census import LineCensus

census = LineCensus.load(f"{ROOT}/census/dndz_census_z0p5_3p0.npz")
area = BaseCatalog(f"{ROOT}/base").footprint.area_deg2

PRIMARY = [1.0e-16, 1.5e-16]                # erg/s/cm^2
SECONDARY = [None, 8.0e-17, 4.0e-17]        # None: one limit for every line
LINES = ["Ha", "OIII_5007", "Hb", "OII_3727"]

z_edges = census.z_edges
z = 0.5 * (z_edges[1:] + z_edges[:-1])
color = dict(zip(SECONDARY, plotting.CATEGORICAL))
label = lambda s: "no secondary limit" if s is None else f"secondary limit {s:.0e}"

print(f"{len(census.steps)} slices, {z_edges[0]:.1f} < z < {z_edges[-1]:.1f}, {area:.1f} deg2")
print("limits (primary, secondary):", census.limits)
```

## dN/dz

Rows: primary limit. Columns: line. Colours: secondary limit. Solid: all galaxies with the line detected;
dashed: single-line; dotted: multi-line.

```python
style = {"all": "-", "single": "--", "multi": ":"}
fig, axes = plt.subplots(len(PRIMARY), len(LINES), figsize=(16, 7.6), sharex=True, sharey=True, constrained_layout=True)
for row, primary in zip(axes, PRIMARY):
    for ax, line in zip(row, LINES):
        for s in SECONDARY:
            for which, ls in style.items():
                ax.step(z, census.dndz(line, primary, area, which, secondary=s), where="mid", color=color[s], ls=ls,
                        lw=1.7 if which == "all" else 1.2, label=f"{label(s)}, {which}")
        ax.set_yscale("log")
        ax.set_ylim(10, 5e4)
        ax.set_xlim(z_edges[0], z_edges[-1])
        ax.set_title(f"{line}, primary limit {primary:.1e}", loc="left")
    row[0].set_ylabel("dN/dz per deg$^2$")
for ax in axes[-1]:
    ax.set_xlabel("redshift")
handles, labels = axes[0, 0].get_legend_handles_labels()
fig.legend(handles, labels, loc="outside lower center", ncol=3, frameon=False)
plt.show()
```

## Single-line fraction of each line's emitters

Fraction of the galaxies with the line detected that have no other detected line.

```python
fig, axes = plt.subplots(1, len(PRIMARY), figsize=(13, 4.4), sharey=True, constrained_layout=True)
for ax, primary in zip(axes, PRIMARY):
    for line in ["Ha", "OIII_5007"]:
        for s, ls in zip(SECONDARY, ["-", "--", ":"]):
            ax.step(z, census.single_fraction(line, primary, secondary=s), where="mid", color=plotting.feature_color(line),
                    ls=ls, label=f"{line}, {label(s)}")
    ax.set_ylim(0, 1)
    ax.set_xlim(z_edges[0], z_edges[-1])
    ax.set_xlabel("redshift")
    ax.set_title(f"primary limit {primary:.1e}", loc="left")
axes[0].set_ylabel("single-line fraction")
axes[-1].legend(frameon=False, loc="center left", bbox_to_anchor=(1.0, 0.5))
plt.show()
```

## Reading the figures

- The secondary limit does not change which galaxies are detected. It moves galaxies from single-line to
  multi-line, and adds the faint lines of detected galaxies to the counts of those lines.
- **Hα** is almost always the brightest line where it is inside the bandpass (z < 1.94), so its total counts do
  not depend on the secondary limit. Its single-line fraction does: with 4e-17 nearly every Hα emitter has a
  second line.
- **[O III] 5007** at z > 1.94, where Hα has left the bandpass, is single-line without a secondary limit and
  multi-line with 4e-17, because [O III] 4959 (flux ratio 1 : 3.04) and Hβ are then detected.
- A secondary limit of 4e-17 is 2.6σ for a 6.5σ limit of 1e-16. The counts assume that such a line is usable
  once another line has anchored the galaxy; they contain no noise.

Other numbers come from the same object, for example
`census.integrated("Ha", 1.0e-16, area, "single", secondary=4.0e-17, z_range=(1.0, 1.9))`.
